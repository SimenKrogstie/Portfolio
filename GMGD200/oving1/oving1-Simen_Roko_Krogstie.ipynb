## **Obligatorisk øving 1**


```python
from geodesy import dms2rad, geod2ECEFs, ECEF2geods
from numpy import sqrt, cos

# Månens posisjon i ECEF koordinater
X_månen = -278712121            # meter
Y_månen = 248335489             # meter
Z_månen = 3199372               # meter

# Kennedy Space Center sin posisjon i geodetiske koordinater
lat_KSC = dms2rad(28, 36, 0)   # radianer
lon_KSC = -dms2rad(80, 36, 0)   # radianer

# Jordradiusen
R = 6371e3                      # meter
```

#### **Oppgave 1**

**a)**

```python
X_KSC, Y_KSC , Z_KSC = geod2ECEFs(R, lat_KSC, lon_KSC, 0)

dX = X_månen - X_KSC 
dY = Y_månen - Y_KSC
dZ = Z_månen - Z_KSC
d = sqrt(dX**2 + dY**2 + dZ**2)
print(f'Avstanden fra Kennedy Space Center til månen er: {d/1e3:.2f} km.')
```

**b)**

```python
lat_m, lon_m, h_m = ECEF2geods(R, X_månen, Y_månen, Z_månen)
print(f'Månen befinner seg {h_m/1e3:.2f} km over jorden.')
```

**c)**

```python
t = 24 * 4          # timer
v = d / (t * 1e3)   # km/h
print(f"Apollo 11 holder en gjennomsnittsfart på {v:.2f} km/h på sin reise til månen.")
```

#### **Oppgave 2**

**a)**

```python
# Lengdegrad til Hornet
lon_Hornet = -dms2rad(169, 54, 0)   # radianer
lon_KSC = -dms2rad(80, 36, 0)       # radianer

delta_lon = lon_KSC - lon_Hornet

d = delta_lon * R * cos(lon_Hornet)
print(f'Apollo 11 lander {d/1e3:.3f} km fra Kennedy Space Center.')
```
