```python
from numpy import log, tan, pi

def Mercator(R, lat, lon):
    """
    Projiserer geodetiske koordinater til Mercatorkoordinater.
    """
    x = R * log(tan(lat/2 + pi/4))
    y = R * lon
    return x, y
```

**d)**

```python
from geodesy import dms2rad

# Jordradius
R = 6371e3                          # meter

# Bredde- og lengdegrad
lat = dms2rad(59, 39, 37.19822)     # radianer
lon = dms2rad(10, 46, 54.18533)     # radianer

# Projiserer posisjonen til Mercatorkoordinater
x, y = Mercator(R, lat, lon)        
print(f"Mercatorkoordinatet for breddegraden: {x:.5f} m")
print(f"Mercatorkoordinatet for lengdegraden: {y:.5f} m")
```
