```python
# Task 4: Read and tabulate data

with open('norway_municipalities_2017.csv') as file:
    lines = file.readlines()[1:]
        
population_district = {}

for i in range(0,len(lines)):
    district = lines[i].split(",")[1]
    population = int(lines[i].split(",")[2])
    population_district[district] = population_district.get(district,0) + population
    
alphabetic = sorted(population_district)
for i in alphabetic:
    print("{:<40} {:<40}".format(i, population_district[i]))
    
sorted_ = sorted(population_district.items(), key= lambda x:x[1], reverse= True)


#Task 5: Plot data

import matplotlib.pyplot as plt
import numpy as np

figure_ = plt.figure(figsize = (10, 5))

district = []
population = []
for i in sorted_:
    district.append(i[0])
    population.append(i[1])
    
plt.bar(district,population)
plt.xticks(rotation=90)
plt.xlabel("District")
plt.ylabel("Population")
plt.title("Population in districts")
plt.show()
```
