```python
# Simen Roko Krogstie og Daniel Fjeld Nyen

#Task 1


student_list = []


with open('student_info.txt', 'r') as file:
    file.readline()
    lines = file.readlines()

    for line in lines:
        line = line.replace(':', ',')
        x = line.split(',')
        name, age, phone_number = x

        student_dict = {
            'name': name.strip(),
            'age': int(age.strip()),
            'phone_number': phone_number.strip()}
        
        student_list.append(student_dict)


for student in student_list:
    print(student)

```
