```python
# Task 1 - Counting votes

import pandas as pd

def voting_distribution(filename, number_of_parties=27):
    df = pd.read_csv(filename, delimiter=';').groupby(by="Partikode", as_index=False)["Antall stemmer totalt"].sum()
    df_sorted = df.sort_values(by="Antall stemmer totalt", ascending=False)

    tot_votes = df["Antall stemmer totalt"].sum()

    df_sorted["Prosent stemmer"] = df_sorted["Antall stemmer totalt"]/tot_votes * 100  
    df_sorted["Prosent stemmer"] = df_sorted["Prosent stemmer"].map('{:.2f}'.format) 
    df_sorted["Prosent stemmer over 4%"] = df_sorted["Prosent stemmer"].astype(float) > 4.00

    table = pd.DataFrame(df_sorted)  

    display(table.iloc[:int(number_of_parties)])

voting_distribution('2021-09-14_party distribution_1_st_2021.csv', 3)
```

```python
# Task 2 - Regex

import re

sentences = ["Ali and Per and friends.", "Kari and Joe know each other.", "James has known Peter since school days."]

print("      Friendships      ")
print("-----------------------")
for sentence in sentences:
    names = re.findall('([A-Z][a-z]+)', sentence)
    
    print(f'{names[0].capitalize():>10} - {names[1].capitalize():<14}')   
```
