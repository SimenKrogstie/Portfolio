```python
import numpy as np
from numpy import pi, arctan
from sympy import Matrix, sqrt

from scipy.stats import chi2, t
```

```python
# Gitte punkter
X216 = 6614767.984
Y216 = 596292.880
X223 = 6612254.730
Y223 = 598328.393

# Ukjente foreløpige punkter
X620 = 6612246.30
Y620 = 595716.10
X827 = 6615189.28
Y827 = 598276.50

# Avstander
d620_216 = 2586.829
d620_223 = 2612.290
d827_223 = 2935.021
d827_216 = 2027.884
d_obs = np.array([d620_216, d620_223, d827_223, d827_216])

# Retninger
R620_216 = 0
R620_827 = 31.2672
R620_223 = 85.4799
R827_223 = 0
R827_620 = 46.7059
R827_216 = 87.8027
r_obs = np.array([R620_216, R620_827, R620_223, R827_223, R827_620, R827_216])

# Koordinatdifferanse: delta x
DX620_216 = (X216-X620)
DX620_827 = (X827-X620)
DX620_223 = (X223-X620)
DX827_223 = (X223-X827)
DX827_620 = (X620-X827)
DX827_216 = (X216-X827)
D_x = np.array([DX620_216, 
                DX620_827, 
                DX620_223, 
                DX827_223, 
                DX827_620, 
                DX827_216])

# Koordinatdifferanse: delta y
DY620_216 = (Y216-Y620)
DY620_827 = (Y827-Y620)
DY620_223 = (Y223-Y620)
DY827_223 = (Y223-Y827)
DY827_620 = (Y620-Y827)
DY827_216 = (Y216-Y827)
D_y = np.array([DY620_216, 
                DY620_827, 
                DY620_223, 
                DY827_223, 
                DY827_620, 
                DY827_216])

# Avstander: D_0
D620_216 = np.sqrt(DX620_216**2 + DY620_216**2)
D620_827 = np.sqrt(DX620_827**2 + DY620_827**2)
D620_223 = np.sqrt(DX620_223**2 + DY620_223**2)
D827_223 = np.sqrt(DX827_223**2 + DY827_223**2)
D827_620 = np.sqrt(DX827_620**2 + DY827_620**2)
D827_216 = np.sqrt((DX827_216**2) + (DY827_216**2))
D_0 = np.array([D620_216, D620_827, D620_223, D827_223, D827_620, D827_216])

Matrix(D_0)
```

```python
# ax: retning
const = 200/pi

ax_r = const * (D_y/D_0**2)
Matrix(ax_r)
```

```python
# ay: retning
ay_r = -const * (D_x/D_0**2)
Matrix(ay_r)
```

```python
# ax: avstand
ax_d = -(D_x/D_0)
Matrix(ax_d)
```

```python
# ay_avstand
ay_d = -(D_y/D_0)
Matrix(ay_d)
```

```python
# Designmatrisen: A
A = np.array([[-1, 0,  ax_r[0],  ay_r[0] ,   0     ,   0     ],
             [-1,  0,  ax_r[1],  ay_r[1] , -ax_r[1], -ay_r[1]],
             [-1,  0,  ax_r[2],  ay_r[2] ,   0     ,   0     ],
             [ 0, -1,    0    ,    0     , ax_r[3] , ay_r[3] ],
             [ 0, -1, -ax_r[4],  -ay_r[4], ax_r[4] , ay_r[4] ],
             [ 0, -1,    0    ,    0     , ax_r[5] , ay_r[5] ],
             [ 0,  0,  ax_d[0],  ay_d[0] ,   0     ,   0     ],
             [ 0,  0,  ax_d[2],  ay_d[2] ,   0     ,   0     ],
             [ 0,  0,    0    ,    0     , ax_d[3] , ay_d[3] ],
             [ 0,  0,    0    ,    0     , ax_d[5] , ay_d[5] ],
             ])
Matrix(A) 
```

```python
# Konstantleddsvektoren: l
# Retninger
phi = arctan(D_y/D_x)*(200/pi)
z0 = phi[0]
z1 = phi[3]

l1 = r_obs + z0- phi
l2 = r_obs + z1 - phi

l_r = Matrix([l1[0], l1[1], l1[2], l2[3], l2[4], l2[5]])

# Avstander
l_d = Matrix([[d_obs[0] - D_0[0]], [d_obs[1] - D_0[2]], [d_obs[2] - D_0[3]], [d_obs[3] - D_0[5]]])

l = l_r.col_join(l_d)
l
```

```python
# # Vektsmatrisen: P
# sigma = 0.0005
# P = np.zeros([10,10])
# p = 1
# np.fill_diagonal(P,p)
# Matrix(P)
```

```python
# Vektsmatrisen: P
s_0 = 0.0005

# Retninger
p_r = np.ones(6)
P_r = np.diag(p_r)         

# Avstander
s_d7 = 0.003 + 0.002 * (D620_216/1000)
s_d8 = 0.003 + 0.002 * (D620_223/1000)
s_d9 = 0.003 + 0.002 * (D827_223/1000)
s_d10 = 0.003 + 0.002 * (D827_216/1000)
s_d = np.array([s_d7, s_d8, s_d9, s_d10])
p_d = (s_0**2) / (s_d**2)
P_d = np.diag(p_d)


P = np.zeros([10, 10])
P[:6, :6] = P_r
P[6:, 6:] = P_d
Matrix(P)
```

```python
# Normalligningene
At = np.transpose(A)
N = At@P@A
h = At@P@l
dx = np.linalg.inv(N)@h


Matrix(dx)
```

```python
# Matriser for oppgitte koordinater
T216 = np.array([6614767.984, 596292.880])
T223 = np.array([6612254.730, 598328.393])
T620 = np.array([6612246.300, 595716.100])
T827 = np.array([6615189.280, 598276.500])
x0 = np.array([[T216], [T223], [T620], [T827]])

x = np.array([[T620[0]+dx[2]], [T620[1]+dx[3]], [T827[0]+dx[4]], [T827[1]+dx[5]]])
Matrix(x)
```

## Deloppgave b)

```python
# Utjevningskorreksjoner
v = A@dx-l
Matrix(v)
```

```python
# Standaradavvik vektsenheten
n = 10
e = 6

s0 = sqrt((np.transpose(v)@P@v)/(n-e))
s0
```

```python
# Standaradavvik på de estimerte parametre
# Kofaktormatrise Q_xx = N^-1
Qxx = np.linalg.inv(N)

s_xi = s0 * np.sqrt(np.diag(Qxx))
s_xi
```
